```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import  r2_score
from sklearn.model_selection import train_test_split
```

```python
data=pd.read_csv("phones_data.csv")
data
```

```python
x2=data.fillna(data['memory_size'].mean())
x2
```

```python
x=data.loc[1:20,'memory_size']
x1=x.fillna(x.mean())
x1
```

```python
x2=data.loc[1:20,'battery_size']
x2
```

```python
X=np.column_stack((x1,x2))
X
```

```python
y=data.loc[1:20,'highest_price']
y=y.fillna(y.mean())
y=np.array(y).reshape(-1,1)
y
```

```python
model=LinearRegression()
model.fit(X,y)
```

```python
batter_size=float(input("Enter your batter size:"))
memory_size=float(input("Enter your memory size:"))
inputss=np.column_stack((batter_size,memory_size))
print(model.predict(inputss))
```

```python
x_test=X
y_test=y
y_pred=model.predict(x_test)
accuracy=r2_score(y_test,y_pred)
print(accuracy*100)
```

```python
import matplotlib.pyplot as plt
fig=plt.figure()
ax=fig.add_subplot(121,projection='3d')
ax.scatter(data.loc[:,'battery_size'],data.loc[:,'memory_size'],data.loc[:,'highest_price'])
plt.title("battery vs memory vs price")
ax.set_xlabel('battery')
ax.set_ylabel('memory')
ax.set_zlabel('price')
plt.show()
```
