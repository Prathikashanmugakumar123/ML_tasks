```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
```

```python
data=pd.read_csv("Experience-Salary.csv")
data
```

```python
x=data[['exp(in months)']]
x
```

```python
y=data['salary(in thousands)']
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
model=LinearRegression() #model
model.fit(x_train,y_train)
```

```python
exp=int(input("Enter your experience:"))
experience=[[exp]]
print("predicted salary",model.predict(experience))
```

```python
y_pred=model.predict(x_test)    #x_test=5   = 40
```

```python
from sklearn.metrics import r2_score
acc=r2_score(y_test,y_pred)
#             50   ,model answer =40
print("Accuracy",acc*100)
```

```python
import matplotlib.pyplot as plt
plt.scatter(x,y)# dotted points
plt.plot(x_test,y_pred,color='red')#line
plt.xlabel("Experience")
plt.ylabel("Salary")
plt.title("Experience vs Salary")
plt.show()
```
