```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
url = "https://www.geeksforgeeks.org/artificial-intelligence/artificial-intelligence-vs-machine-learning-vs-deep-learning/"
```

```python
response = requests.get(url)
```

```python
soup = BeautifulSoup(response.text, "html.parser")
```

```python
table = soup.find("table")

rows = table.find_all("tr")
```

```python
data = []

for row in rows:
    cols = row.find_all(["td", "th"])
    cols = [c.text.strip() for c in cols]
    data.append(cols)
```

```python
df = pd.DataFrame(data[1:], columns=data[0])
df
```

```python
df.to_csv("dataset.csv", index=False)

print("CSV created successfully!")
print(df)
```

```python
import numpy as np
from statsmodels.discrete.count_model import ZeroInflatedPoisson

# Training data
age = np.array([
    22, 25, 28, 30, 32,
    35, 38, 40, 45, 50,
    52, 55, 58, 60, 65
])

claims = np.array([
    0, 0, 1, 0, 2,
    1, 0, 2, 3, 0,
    2, 4, 1, 3, 5
])

# Check sizes
print("Age:", len(age))
print("Claims:", len(claims))

# Input data
X = age.reshape(-1, 1)

# Train model
model = ZeroInflatedPoisson(
    endog=claims,
    exog=X,
    exog_infl=X
)

result = model.fit(disp=0)

print("\nModel trained successfully!")

# User input
user_age = float(input("\nEnter user age: "))

# New user data
new_user = np.array([[user_age]])

# Prediction
prediction = result.predict(
    exog=new_user,
    exog_infl=new_user
)

print("Predicted number of claims:", round(prediction[0], 2))
```
