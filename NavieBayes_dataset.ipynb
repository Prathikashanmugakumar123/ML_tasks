```python
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.preprocessing import LabelEncoder
```

```python
data=pd.read_csv("email.csv")
data
```

```python
emails=data["text"]
y=data["spam"]
```

```python
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(emails)
```

```python
model = MultinomialNB()
model.fit(X, y)
```

```python
mail=input("Enter your mail words:")
test_email = [mail]
test_vec = vectorizer.transform(test_email)

pred = model.predict(test_vec)
print(pred)
```

```python
if pred==0:
  print("not spam")
else:
  print("spam")
```

```python
from sklearn.naive_bayes import GaussianNB
students_details=pd.read_csv("Pass.csv")
students_details
```

```python
sub1=students_details.loc[1:50,"subject1"]
x1=np.array(sub1)
x1
```

```python
sub2=students_details.loc[1:50,"subject2"]
x2=np.array(sub2)
x2
```

```python
add=x1+x2
total=np.array(add)
total
```

```python
X=np.array(total).reshape(-1,1)
X
```

```python
y_text=students_details.loc[1:50,"Pass"]
y_text
```

```python
le = LabelEncoder()
y = le.fit_transform(y_text)
y
```

```python
model = GaussianNB()
model.fit(X, y)
```

```python
mark=int(input("Enter your mark:"))
test_mark=np.array(mark).reshape(-1,1)
test_mark
answer=model.predict(test_mark)
print(answer)
print(le.inverse_transform(answer))
```

```python
import numpy as np
from sklearn.naive_bayes import BernoulliNB
from sklearn.preprocessing import LabelEncoder
```

```python
ds=pd.read_csv("food_set.csv")
ds
```

```python
x1=ds["Spicy"]
x2=ds["Cheese"]
X=np.column_stack((x1,x2))
X
```

```python
y_text=ds["Likes Food"]
y_text
```

```python
le = LabelEncoder()
y = le.fit_transform(y_text)
y
```

```python
model=BernoulliNB()
model.fit(X,y)
```

```python
spicy=int(input("enter your answer(1-yes,no-0):"))
cheese=int(input("enter your answer(1-yes,no-0):"))
test_food=np.array([spicy,cheese]).reshape(1,-1)
test_food
answer=model.predict(test_food)
print(answer)
print(le.inverse_transform(answer))
```
